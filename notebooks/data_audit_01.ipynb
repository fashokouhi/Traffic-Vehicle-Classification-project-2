```python
from pathlib import Path
import pandas as pd

root = Path(train_dataset.root)
# نگه داشتن مسیر اصلی دیتاست همان دیتای ترکیب  شده

all_files = {
    path.stem: path
    # نام فایل بدون پسوند jpg
    for path in root.rglob("*")
    # تمام فایل ها از جمله پوشه ها و زیرپوشه های فایل اصلی (ریشه) را پیداکن
    if path.is_file() and path.suffix.lower() in [".jpg", ".jpeg", ".png"]
}

found = []
not_found = []

for class_name, image_ids in suspicious.items():

    for image_id in image_ids:

        image_id = str(image_id)

        if image_id in all_files:
            path = all_files[image_id]

            found.append({
                "filename": path.name,
                "current_label": path.parent.name,
                "expected_label": class_name,
                "path": str(path)
            })

        else:
            not_found.append((class_name, image_id))

print("Found:", len(found))
print("Not found:", len(not_found))
```
